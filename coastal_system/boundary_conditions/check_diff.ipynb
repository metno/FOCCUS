```python
import xarray as xr
import numpy as np
#import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [12, 8]
plt.rcParams['font.size'] = 18
```

```python
# dropping everything but salt_obc and temp_obc
drop_vars = ['AKs', 'AKt', 'AKv', 'Cs_r', 'Cs_w', 'DU_avg1', 'DU_avg2', 'DV_avg1', 'DV_avg2', 'h', 'hc', 'salt', 'temp', 'u', 'u_obc', 'ubar', 'ubar_obc', 'v', 'v_obc', 'vbar', 'vbar_obc', 'zeta', 'zeta_obc']
```

```python
year = 2024

for month in ['01', '02', '03', '04', '05', '06', '07', '08', '09', '10', '11', '12']:
    day = 10
    dir = f'https://thredds.met.no/thredds/dodsC/foccus/'

    ds0 = xr.open_dataset(dir + f'norshelf-ana_fwd_{year}{month}{day}T00Z_outer0.nc', drop_variables=drop_vars)
    ds1 = xr.open_dataset(dir + f'norshelf-ana_fwd_{year}{month}{day}T00Z_outer1.nc', drop_variables=drop_vars)

    #-----
    # Variables with ending "_obc" contains 2D boundaries / slices (+ time dimension)
    # which are selected from the index 'boundary'
    ib = {'south': 0, 'east': 1, 'north': 2, 'west': 3}
    salt_west_obc0 = ds0['salt_obc'].isel(ocean_time=0, boundary=ib['west'], obc_adjust=0).values
    salt_west_obc1 = ds1['salt_obc'].isel(ocean_time=0, boundary=ib['west'], obc_adjust=0).values

    diff = salt_west_obc0 - salt_west_obc1

    #-----
    print('--------------------------------------')
    print(f'{year}-{month}-{day} at west obc:')
    print(f'Max:  {diff.max():.4f}')
    print(f'Min:  {diff.min():.4f}')
    print(f'Mean: {diff.mean():.4f}')
```

```python
vars = []
for var in ds0.data_vars:
    vars.append(var)
print(f'Variables in the dataset: {vars}')
```
